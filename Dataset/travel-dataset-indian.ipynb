```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("/kaggle/input/travel-dataset-guide-to-indias-must-see-places/Top Indian Places to Visit.csv")
data
```

```python
data.head() #first 5 rows
```

```python
data.tail() #last 5 rows
```

# Distribution of Zone

```python
data["Zone"].value_counts().plot(kind ='bar' , color="navy")
plt.title("distribution of Zone")
plt.xlabel("Zone")
plt.ylabel("Count")
plt.show()
```

# Distribution of States

```python
data["State"].value_counts().plot(kind ='bar' , color="navy")
plt.title("distribution of Satats")
plt.xlabel("State")
plt.ylabel("Count")
plt.show()
```

# City


```python
city=data["City"].value_counts()
city
```

```python
most_10_city=(data["City"].value_counts()).head(10)
most_10_city.plot(kind='bar' ,color="seagreen")
plt.title("Most 10 cities visited")
plt.show()
```

# Count of Landmarks

```python
places=data["Name"].value_counts()
places
```

```python
place_10_most=places.head(10)
place_10_most.plot(kind='bar' , color='tan')
plt.title("Names of landmarks distribution")
plt.show()
```

# Classification of the place

```python
types=data["Type"].value_counts()  
types
```

```python
#most 10  types visited

types_10_most=types.head(10)
types_10_most.plot(kind='bar' , color='gold')
plt.xlabel("Type")
plt.ylabel("Count")
plt.title("Most 10 types visited")
plt.show()
```

# Most 10 cities VS. Significance

```python
plt.figure(figsize=(20,20))
City_Significance=data.groupby(["City" ,"Significance"]).size().unstack()
(City_Significance.head(10)).plot(kind='bar' , stacked=True)
plt.title("Most 10 cities VS. Significance ")
plt.xlabel("City")
plt.legend(loc=(1,0))
plt.show()
```

# use DSLR cameras

```python
data["DSLR Allowed"].value_counts().plot(kind='bar' , color=["tan" , "bisque"])
plt.title("DSLR Allowed Or Not")
plt.xlabel("DSLR")
plt.ylabel("Count")
plt.show()
```

# Scatter Plot

```python
sns.scatterplot(data=data ,x="time needed to visit in hrs" ,y="Number of google review in lakhs" )
plt.title("Time Needed to visit in hrs VS. Number of google review in lakhs")
plt.show()
```

```python
sns.scatterplot(x="time needed to visit in hrs",y="Google review rating" ,data=data)
plt.title("Time Needed to visit in hrs VS. Google review rating")
plt.grid()
plt.show()
```

```python
sns.set_style("darkgrid")
sns.relplot(x="Google review rating",y="Entrance Fee in INR" , data=data , kind="scatter",hue="Significance")

plt.show()
```

# barplot between Number of google review in lakhsand Significance

```python
sns.catplot(x="Significance",y="Number of google review in lakhs", data=data,kind="bar")
plt.xticks(rotation=90)
plt.show()
```

```python
sns.catplot(x="Zone",y="time needed to visit in hrs",data=data,kind="box")
plt.show()
```
